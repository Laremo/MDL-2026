### Paso 3: Implemnentar Data Aumentation

#### a) Ejemplo de "rotation range". Es un valor en grados (0-180) que indica dentro del cual se puede rotar imágenes al azar

```python
# Datagen = ImageDataGenerator(rotation_range=180)
```

```python
# from google.colab import files
from tensorflow.keras.preprocessing import image

# uploaded=files.upload()
# for fn in uploaded.keys()
path = test_dogs_dir + "/dog.1669.jpg"
img=image.load_img(path)
data = img_to_array(img)
samples = expand_dims(data, 0)

# example of "rotation_range"
datagen = ImageDataGenerator(rotation_range=180)

it = datagen.flow(samples, batch_size=1)
for i in range(6):
    pyplot.subplot(230 + 1 + i)
    batch = next(it)
    image = batch[0].astype('uint8')
    pyplot.imshow(image)
pyplot.show()
```

#### b) Ejemplo de "width_shift_range". Son rangos (como una fracción del ancho y la altura total) dentro de los cuales se pueden transladar las imágenes al azar verticalmente u horizontalmente

```python
# from google.colab import files
from tensorflow.keras.preprocessing import image

# uploaded=files.upload()
# for fn in uploaded.keys()
path = test_dogs_dir + "/dog.1511.jpg"
img=image.load_img(path)
data = img_to_array(img)
samples = expand_dims(data, 0)

# example of "rotation_range"
datagen = ImageDataGenerator(width_shift_range=0.8)
# datagen = ImageDataGenerator(height_shift_range=0.5)

it = datagen.flow(samples, batch_size=1)
for i in range(6):
    pyplot.subplot(230 + 1 + i)
    batch = next(it)
    image = batch[0].astype('uint8')
    pyplot.imshow(image)
pyplot.show()
```

#### c) Ejemplo de "shear_range". Sirve para aplicar transformaciones de corte al azar

```python
# from google.colab import files
from tensorflow.keras.preprocessing import image

# uploaded=files.upload()
# for fn in uploaded.keys()
path = test_dogs_dir + "/dog.1511.jpg"
img=image.load_img(path)
data = img_to_array(img)
samples = expand_dims(data, 0)

datagen = ImageDataGenerator(shear_range=60)

it = datagen.flow(samples, batch_size=1)
for i in range(6):
    pyplot.subplot(230 + 1 + i)
    batch = next(it)
    image = batch[0].astype('uint8')
    pyplot.imshow(image)
pyplot.show()
```

#### d) Ejemplo de "zoom_range" Sirve para aplicar zoom aletaroio dentro de las imagenes

```python
# from google.colab import files
from tensorflow.keras.preprocessing import image

# uploaded=files.upload()
# for fn in uploaded.keys()
path = test_dogs_dir + "/dog.1511.jpg"
img=image.load_img(path)
data = img_to_array(img)
samples = expand_dims(data, 0)

datagen = ImageDataGenerator(zoom_range=0.9)

it = datagen.flow(samples, batch_size=1)
for i in range(6):
    pyplot.subplot(230 + 1 + i)
    batch = next(it)
    image = batch[0].astype('uint8')
    pyplot.imshow(image)
pyplot.show()
```

#### e) Ejemplo de "horizontal_flip" Sirve para volter aleatoriamente la imagen horizontalmente

```python
# from google.colab import files
from tensorflow.keras.preprocessing import image

# uploaded=files.upload()
# for fn in uploaded.keys()
path = test_dogs_dir + "/dog.1511.jpg"
img=image.load_img(path)
data = img_to_array(img)
samples = expand_dims(data, 0)

datagen = ImageDataGenerator(horizontal_flip=True)

it = datagen.flow(samples, batch_size=1)
for i in range(6):
    pyplot.subplot(230 + 1 + i)
    batch = next(it)
    image = batch[0].astype('uint8')
    pyplot.imshow(image)
pyplot.show()
```

#### f) Ejemplo de "fll_mode" Esta estrategia se utilza para llenar los pizeles recien creados que pueden aparecer despues de una de las transformaciones anteriores

```python
# from google.colab import files
from tensorflow.keras.preprocessing import image

# uploaded=files.upload()
# for fn in uploaded.keys()
path = test_dogs_dir + "/dog.1511.jpg"
img=image.load_img(path)
data = img_to_array(img)
samples = expand_dims(data, 0)

datagen = ImageDataGenerator(fill_mode='nearest')

it = datagen.flow(samples, batch_size=1)
for i in range(6):
    pyplot.subplot(230 + 1 + i)
    batch = next(it)
    image = batch[0].astype('uint8')
    pyplot.imshow(image)
pyplot.show()
```
