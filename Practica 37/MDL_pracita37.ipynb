## 6.Periograma
Aplicamoes el peridiograma y ponemos logaritmo=no, para que no suavice la serie. El peridiograma nos permite descubrir losp eriodos mas importantes, en este caso los tre periodos mas importantes que obtuvimos mayor numero de ventas

```r
res <- stl(lbeer, log="no")
res$freq
```

```r
## res me da la frecuencia  y el espectro, que es lo que se grafica en el grafico del peridiograma. 
### El peridfiograma grafica la frecuencia en eje x, y el espectro en eje y.
## Tendo que descibrir cual aclano es el mas importante, el que tiene el espectro mas alto.
order(res$spec, res$req, decreasing=TRUE)
```

```r
## Los tres periodos mas importantes, estan en la posicion 40, 2, y 120.
# posterioremente descubrimos sus valores

max1 <- res$freq[40]
max1
```

```r
max2 <- res$freq[2]
max2
```

```r
max3 <- res$freq[120]
max3
```

```r
### El primer pico ocurre en 0.0041
abline(v=max1, col="red", lyx="dotted")

periodo1 <- 1/max1
periodo1
```

```r
## El segundo pico ocurre en 0.08

abline(v=max2, col="blue", lyx="dotted")
periodo2 <- 1/max2
periodo2
```

```r
## El tercer pico ocurre en 0.25, en el maximo 3

abline(v=max3, col="green", lyx="dotted")
```
