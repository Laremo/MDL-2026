## Practica 31 Series de tiempo
#### Hoy en día las organizaciones que utilizan internet requieren conocer el comportamiento futuro de ciertos fenómenos con el fin de planificar o prevenir, esde cri, se utilizarn para preder lo que ocurrirá con una variable en el futuro a parti del comportamiento de esa variable en el pasado.

#### Se llama series de tiempo a un conjunto d eobservacione sobre valores que toma una variable (cuantitativa) en diferentes momento en el tiempo

#### Una seride de tiempo es una colección o conjunto de mediciones de cirto fenómeno o experimento registrados secuencialmente en el tiempo, en forma equispaciada (a intervalos de tiempo iguales)

### En esta practica implementaremos varias series de tiempo utilizando el dataset sockdata.csv, referente al mercado de valores (yahoo finanzas https://es-us.finanzas.yahoo.com/)




## Paso 1: Cargar los datos del mercado de valores

```r
stock.data <- read.csv("stockdata.csv", header = T, dec=",", sep = ";")
head(stock.data)
summary(stock.data)
```

## Paso 2: Gráfica series de tiempo para el volumen

```r
plot(stock.data[,5], type="l")
```

## Paso 3: Grafica de series de tiempo para el precio de cierre

```r
plot(stock.data$close, type="l")
```

## Paso 4: Gráfica de series de tiempo para el precio de apertura 

```r
plot(stock.data[,2], type="o")
plot(stock.data$open, type="o")
```

## Paso 5: Gráfica series de tiempo para el precio más bajo y más alto (rojo)

```r
plot(stock.data$low, type="l")
lines(stock.data$high, type="l", col="red")
```

## Paso 6: Gráfica series de tiempo para el precio de cierre con un grafico diferente

```r
plot(stock.data$close, type="l", lwd=2, col="red", xlab="time", ylab="closing values", main="STOCK DATA", ylim=c(0,60))
```
